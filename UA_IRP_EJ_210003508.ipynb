## Methodology <a class="anchor" id="methodology"></a>

> Describe the methods used in your research, including data sources, collection procedures, analysis techniques, and any relevant software tools. Include code cells for data loading, preprocessing, and key steps in your analysis pipeline.

This project investigates spatial patterns of crime in London by developing geodemographic and crime-based classifications at the Output Area (OA) level. The analysis combines 2024 police crime data with 2021 Census demographic data, applying various spatial and statistical techniques to uncover clusters of similar areas. This methodology outlines the data sources, preprocessing steps, geospatial integration, and the classification methods used.

The project was implemented in Python using a suite of libraries including pandas, geopandas, matplotlib, seaborn, plotly, shapely, folium, leafmap, scikit-learn, mapclassify, and sodapy. Spatial operations used the WGS84 coordinate reference system (EPSG:4326). All code was written in Jupyter Notebooks to ensure transparency, reproducibility, and seamless integration of spatial and statistical workflows.

By synthesising demographic data into interpretable geodemographic profiles and mapping these against spatial crime patterns, this methodology enables a nuanced exploration of urban inequality, social structure, and crime in London.

### Study Area

London, with a population exceeding nine million and over 33 boroughs, is among the most diverse and densely populated urban centres in the world. Its heterogeneity in terms of ethnicity, housing, employment, and income makes it a compelling case study for spatial analyses of crime and socio-economic inequality. The analysis in this product was conducted at the Output Area (OA) level—the smallest geographical unit used in the UK Census—providing a high-resolution spatial lens for examining urban inequality and localised crime trends. There are 26,369 OAs in Greater London. These units typically contain 100 to 625 residents and offer a suitable spatial scale for geodemographic classification.

### Data Collection and Cleaning 

Three primary data sources were used. 

To prepare the data for analysis, crime CSV files were merged and cleaned. Key columns such as crime type, longitude, latitude, LSOA code, and month were retained. A GeoDataFrame of crime incidents was created in Python using GeoPandas, and each record was spatially joined to its corresponding Output Area polygon based on geographic coordinates.

Census data was preprocessed by selecting percentage-based indicators related to housing, deprivation, health, education, employment, and migration. These variables were normalised using total counts from the relevant OA to generate comparable indicators. The resulting datasets were merged using Output Area codes and spatially joined to the OA shapefile for mapping and analysis.

#### Socio-economic Data

To complement the crime data, socio-economic and demographic indicators were sourced from the 2021 UK Census at the Output Area (OA) level via [NOMIS](https://www.nomisweb.co.uk/sources/census_2021_bulk). The OA level provides the finest spatial resolution available, enabling detailed geodemographic classification across London.

These variables offer essential context for understanding spatial crime patterns and capture a wide range of **risk factors and social determinants** associated with crime — including deprivation, unemployment, housing, education, ethnicity, health, and population density. All variables were converted to percentages to ensure comparability across OAs, and z-score standardisation was applied to normalise the data prior to clustering.

Each Census table was downloaded as a separate CSV and merged using the OA code as a unique identifier. The final dataset supports the development of geodemographic profiles to better understand crime trends at the neighbourhood level.

All percentage-based socio-demographic and crime indicators were standardised using z-score normalisation to ensure comparability across variables and enhance clustering performance. Multicollinearity was addressed by computing Pearson correlation matrices. Variables exhibiting strong correlations (r > 0.7), such as “students” and “routine occupations”, were removed to avoid redundancy and ensure model stability.

### Geodemographic Clustering

K-Means clustering was used to generate geodemographic neighbourhood types based on the transformed SES data. The optimal number of clusters was determined using the Elbow Method and Between-Cluster Sum of Squares (BSS). Once established, clusters were interpreted using radar plots that illustrated the socio-demographic profiles of each group. Clusters were also mapped across London using Folium and Leafmap to explore their spatial distribution interactively.

Geodemographic classification is a particularly powerful tool in this context because it allows for the synthesis of diverse Census and socio-economic data to define meaningful clusters of urban populations. These clusters can capture complex social realities that simple deprivation indices may overlook, providing a more nuanced foundation for examining how different communities experience and are exposed to crime. By using clustering methods such as K-Means and incorporating data on age, ethnicity, education, housing, and employment from the 2021 Census, this study identifies distinct geodemographic profiles across London. These are then spatially compared with crime data to assess patterns of vulnerability and resilience.

### Spatial Crime Analysis

Crime data was aggregated to the Output Area level and linked to each OA’s cluster classification. For each Output Area, crime counts were summarised by type and totalled across all categories. Descriptive statistics were produced, including boxplots comparing crime distributions across clusters, bar charts showing average crime per cluster, and cluster-specific crime profiles.

Statistical tests were conducted to explore relationships between crime and geodemographic types. One-way ANOVA was used to assess whether crime levels significantly differed between clusters. Pearson correlation analysis and heatmaps quantified the association between individual SES indicators and different crime types. Linear regression models were then constructed, with dummy-coded cluster membership used to predict total crime. These models produced coefficient estimates and R² values to assess explanatory power.

### Mapping and Visualisation

Spatial visualisation was a key component of the analysis. Choropleth maps were used to display crime rates and SES indicators using classification schemes including Natural Breaks (Jenks), Equal Intervals, and Quantiles. Interactive maps were developed with hover-enabled overlays showing cluster membership, crime intensity, and variable scores by Output Area. Histograms and boxplots highlighted distributional trends and outliers, while bar plots ranked the top 20 Output Areas by total crime count.

### Limitations

Several limitations must be acknowledged. The Output Area is the smallest geographical unit for which full Census data is available, but some heterogeneity remains within these zones. The K-Means algorithm assumes spherical clusters of equal size, which may oversimplify complex spatial phenomena. Finally, using a single year of crime data captures only a snapshot in time and may not reflect longer-term trends.

#### House Prices and Crime 

In the correlation matrix, the coefficient between total crime and house prices in 2021 (hpmd202103) is 0.1137, indicating a very weak positive relationship. This suggests that areas with higher total crime tend to have slightly higher house prices, but the relationship is so weak that it is practically negligible. In urban contexts like London, this is not surprising: crime rates can be higher in wealthier or more central neighbourhoods (such as Westminster or Camden) due to large numbers of people, nightlife, tourism, and commercial activity, while property prices remain high because the location itself is highly desirable. As a result, total crime counts alone are not strongly predictive of house price levels in London, and more nuanced relationships might be found if specific types of crime are considered separately rather than total crime counts alone. It may therefore be valuable to explore whether different categories of crime (e.g., violent crime, property crime, anti-social behaviour) have stronger or weaker associations with house prices individually.
